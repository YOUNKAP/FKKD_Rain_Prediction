```python
import pandas as pd
```

```python
# Path to your Excel file
excel_path = "FKKD_2024.xlsx"  # 
```

```python
excel_file = pd.ExcelFile(excel_path, engine='openpyxl')
sheet_names = excel_file.sheet_names
```

```python
print(sheet_names)
```

```python
# List to collect each sheet's data
all_data = []
```

```python
for sheet in sheet_names:
    df = pd.read_excel(excel_path, sheet_name=sheet, engine='openpyxl')
    #df["Sheet"] = sheet  # Optional: track source sheet
    df["Hour"] = df["Hour"].astype(str).str.zfill(4)  # Ensure 4-digit strings
    df["Hour_int"] = df["Hour"].str[:2].astype(int)
    df["Minute"] = df["Hour"].str[-2:].astype(int)
    all_data.append(df)
```

```python
# Merge all dataframes
weather_df = pd.concat(all_data, ignore_index=True)
```

```python
weather_df.head()
```

```python
weather_df.tail()
```

```python
weather_df["Datetime"] = pd.to_datetime(weather_df[["Year", "Month", "Day", "Hour_int", "Minute"]].rename(columns={"Hour_int": "Hour"}))
```

```python
#weather_df["Datetime"] = pd.to_datetime(weather_df[["Year", "Month", "Day", "Hour"]])
```

```python
# Sort by datetime if needed
weather_df = weather_df.sort_values(by="Datetime").reset_index(drop=True)
```
