```python
import pandas as pd
import os
from pathlib import Path
```

```python
#! pip install XlsxWriter
```

```python
# Set your CSV directory and output Excel file path
csv_folder = Path("/home/duplex/Desktop/remote/crop_mapping/EO_Africa_OC04/fPAR/vegetation")
output_excel = "merged_output.xlsx"
```

```python
# Create a Pandas Excel writer using XlsxWriter as the engine
with pd.ExcelWriter(output_excel, engine="xlsxwriter") as writer:
    for csv_file in csv_folder.glob("*.csv"):
        sheet_name = csv_file.stem[:31]  # Excel sheet names must be <= 31 characters
        df = pd.read_csv(csv_file)
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"All CSVs merged into {output_excel}")
```
