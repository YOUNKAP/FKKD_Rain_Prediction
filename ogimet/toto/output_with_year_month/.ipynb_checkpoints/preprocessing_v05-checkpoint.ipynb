## DROP RAW

```python
print(len(data))
```

```python
data = data[data["Temp"].astype(str) != "Q1007"]
```

```python
print(len(data))
```

```python
data = data[data["DP"].astype(str) != "NO"]
```

```python
print(len(data))
```

```python
data.describe()
```

```python
data["Visibility"] = pd.to_numeric(data["Visibility"], errors="coerce")
moving_avg = data["Visibility"].rolling(window=3, min_periods=1).mean()
data["Visibility"] = data["Visibility"].fillna(moving_avg)
data["Visibility"] = data["Visibility"].round(0).astype("Int64")
```

```python
data["Visibility"].head()
```

```python
data["Pressure"] = pd.to_numeric(data["Pressure"])
moving_avg = data["Pressure"].rolling(window=3, min_periods=1).mean()
data["Pressure"] = data["Pressure"].fillna(moving_avg)
data["Pressure"] = data["Pressure"].round(0).astype("Int64")
```

```python
data.describe()
```

```python
data["Temp"] = pd.to_numeric(data["Temp"])
moving_avg = data["Temp"].rolling(window=3, min_periods=1).mean()
data["Temp"] = data["Temp"].fillna(moving_avg)
data["Temp"] = data["Temp"].round(0).astype("Int64")
```

```python
data["DP"] = pd.to_numeric(data["DP"])
moving_avg = data["DP"].rolling(window=3, min_periods=1).mean()
data["DP"] = data["DP"].fillna(moving_avg)
data["DP"] = data["DP"].round(0).astype("Int64")
```

```python
data.describe()
```

```python
data.dtypes
```

## Label Encoding for Rain

```python
print(data["Rain"].unique())
```

```python
# Map "yes" → 1, "no" → 0
data["Rain"] = data["Rain"].map({"YES": 1, "NO": 0}).astype("int")
```

```python
print(data["Rain"].unique())
```
