```python
def relative_humidity(temp_c, dew_point_c):
    numerator = math.exp((17.67 * dew_point_c) / (dew_point_c + 243.5))
    denominator = math.exp((17.67 * temp_c) / (temp_c + 243.5))
    rh = 100 * (numerator / denominator)
    return rh
```

```python
# Set your CSV directory and output Excel file path
csv_folder = Path("/home/duplex/Downloads/ogimet/toto/output_with_year_month/2024")
output_excel = "FKKD_2024.xlsx"
```

```python
# Create a Pandas Excel writer using XlsxWriter as the engine
with pd.ExcelWriter(output_excel, engine="xlsxwriter") as writer:
    for csv_file in csv_folder.glob("*.csv"):
        print(len(str(csv_file)))
        sheet_name = csv_file.stem[:31] # Excel sheet names must be <= 31 characters
        sheet_name = str(csv_file)[78:-4]
        print(sheet_name)
        df = pd.read_csv(csv_file)
        Days = []
        Hours = []
        Temp = []
        DP = []
        RH = []
        Wind_Speed =[]
        Pressure = []
        RH = []
        df.to_excel(writer, sheet_name=sheet_name, index=False)
        for i in range(0, len(df)):
            day = df["Time"][i][:2]
            Days.append(day)
            hour = df["Time"][i][2:-1]
            Hours.append(hour)
            temperature = df["Temperature"][i]
            if isinstance(temperature, str) and "/" in temperature:
                temp= int(temperature[:2])
                #Temp.append(temp)
                dp = int(temperature[-2:])
                #DP.append(dp)
                try:
                    Temp.append(temp)
                    DP.append(dp)
                except ValueError:
                    Temp.append(None)
                    DP.append(None)
            else:
                Temp.append(None)
                DP.append(None)
            rh = round(relative_humidity(temp, dp),1)
            RH.append(rh)
            val = df["Wind"][i]
            if pd.notnull(val):
                wind_speed = str(val)[-4:-2]
            else:
                wind_speed = None
            Wind_Speed.append(wind_speed)
            value = df["Sea Level Pressure"][i]
            pressure = str(value)[1:] if pd.notnull(value) else None
            Pressure.append(pressure)
        df["Day"] = Days
        df["Hour"] = Hours
        df["Temp"] = Temp
        df["DP"] = DP
        df["RH"] = RH
        df["Wind Speed"] = Wind_Speed
        df["Pressure"] = Pressure
        valid_data = df[["Year", "Month", "Day", "Hour","Temp", "DP", "RH", "Wind Speed","Visibility", "Rain","Pressure"]]
        valid_data.to_excel(writer, sheet_name=sheet_name, index=False)
print(f"All CSVs merged into {output_excel}")
```
