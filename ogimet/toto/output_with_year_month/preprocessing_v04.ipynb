```python
import pandas as pd
import glob
import os
```

```python
csv_folder = "processed_data/"  
output_file = "FKKD_2021_to_2024_processed.csv"
```

```python
csv_files = sorted(glob.glob(os.path.join(csv_folder, "FKKD_202*.csv")))
```

```python
for csv_file in csv_files :
    print(csv_file)
```

```python
all_data = []
```

```python
for csv_file in csv_files:
    df = pd.read_csv(csv_file)
    #df["SourceFile"] = os.path.basename(csv_file)  
    all_data.append(df)
```

```python
combined_df = pd.concat(all_data, ignore_index=True)
```

```python
combined_df.to_csv(output_file, index=False)
```
