```python
import pandas as pd
import math
```

```python
def relative_humidity(temp_c, dew_point_c):
    numerator = math.exp((17.67 * dew_point_c) / (dew_point_c + 243.5))
    denominator = math.exp((17.67 * temp_c) / (temp_c + 243.5))
    rh = 100 * (numerator / denominator)
    return rh
```

```python
Days = []
Hours = []
Temp = []
DP = []
RH = []
Wind_Speed =[]
Pressure = []
RH = []
```

```python
for i in range(0, len(df)):
    day = df["Time"][i][:2]
    Days.append(day)
    hour = df["Time"][i][2:-1]
    Hours.append(hour)
    temp = int(df["Temperature"][i][:2])
    Temp.append(temp)
    dp = int(df["Temperature"][i][-2:])
    DP.append(dp)
    rh = relative_humidity(temp, dp)
    RH.append(rh)
    val = df["Wind"][i]
    if pd.notnull(val):
        wind_speed = str(val)[-4:-2]
    else:
        wind_speed = None
    Wind_Speed.append(wind_speed)
    value = df["Sea Level Pressure"][i]
    pressure = str(value)[1:] if pd.notnull(value) else None
    Pressure.append(pressure)
```

```python
df["Day"] = Days
```

```python
df["Hour"] = Hours
```

```python
df["Temp"] = Temp
```

```python
df["DP"] = DP
```

```python
df["RH"] = RH
```
